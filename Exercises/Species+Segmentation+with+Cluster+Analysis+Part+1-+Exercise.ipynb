## Import the relevant libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn import preprocessing
sns.set()
```

## Load the data

Load data from the csv file: <i> 'iris_dataset.csv'</i>.

```python
raw_data = pd.read_csv('iris_dataset.csv')
raw_data
```

## Plot the data

For this exercise, try to cluster the iris flowers by the shape of their sepal. 

<i> Use the 'sepal_length' and 'sepal_width' variables.</i> 

```python
plt.scatter(raw_data['sepal_length'],raw_data['sepal_width'])
plt.xlabel('sepal_length')
plt.ylabel('sepal_width')
plt.show()
```

# Clustering (unscaled data)

Separate the original data into 2 clusters.

```python
x = raw_data.copy()
kmeans = KMeans(2)
kmeans.fit(x)
```

```python
clusters = raw_data.copy()
clusters['cluster_pred']=kmeans.fit_predict(x)
```

```python
clusters
```

```python
plt.scatter(clusters['sepal_length'], clusters['sepal_width'],c=clusters['cluster_pred'],cmap='rainbow')
```

# Standardize the variables

Import and use the <i> method </i> function from sklearn to standardize the data. 

```python
x_scaled = preprocessing.scale(x)
x_scaled
```

# Clustering (scaled data)

```python
clusters_scaled = raw_data.copy()
clusters_scaled['cluster_pred']=kmeans.fit_predict(x_scaled)
```

```python
plt.scatter(clusters_scaled['sepal_length'],clusters_scaled['sepal_width'],c=clusters_scaled['cluster_pred'],cmap='rainbow')
```

## Take Advantage of the Elbow Method

### WCSS

```python
wcss = []
c1_num = 10
for i in range(1,c1_num):
    kmeans = KMeans(i)
    kmeans.fit(x_scaled)
    wcss_iter = kmeans.inertia_
    wcss.append(wcss_iter)
```

```python
wcss
```

### The Elbow Method

```python
number_clusters = range(1,c1_num)
plt.plot(number_clusters,wcss)
plt.title('The elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Within-cluster Sum of Squares')
```
