# Deep Learning — Minimal Example (TensorFlow)

The **same** minimal example as the from-scratch NumPy version, rebuilt with **TensorFlow**.
Same idea — generate data, define a linear model, minimize an L2 loss with gradient descent —
but the framework handles the gradients and weight updates automatically.

Companion to `Deep Learning - Minimal Example.ipynb` (the by-hand version). — Waheed

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
print(tf.__version__)
```

```python
observations = 1000

xs = np.random.uniform(low=-10,high=10, size=(observations,1))
zs = np.random.uniform(-10,10, (observations,1))

generated_inputs = np.column_stack((xs,xz))
noise  = np.random.uniform(-1,1, (observations,1))
generated_targets = 2*xs - 3*xz + 5 + noise

np.savez('TF_intro', inputs = generated_inputs, targets = generated_targets)
```
